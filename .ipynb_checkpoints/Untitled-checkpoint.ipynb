```python
import numpy as np
import pandas as pd 
import matplotlib as plt 
```

```python
# read csv
countries = pd.read_csv("World_Development_Indicators\world_bank_data.csv")

def get_country_data(country_name: str):
    result = countries[countries["Country Name"] == country_name].drop(columns = ["Country Name", "Country Code"])
    result = result.transpose()
    result.columns = result.iloc[0]
    return result[1:]

# seperate countries
japan = get_country_data("Japan")
united_states = get_country_data("United States")
china = get_country_data("China")

# list varibles in data (columns)  note: the columns and index are the same for all 3 country dataframes 
print("COLUMNS:")
for col in japan.columns:
    print(col)

# list years in data (index)
print("INDEX:")
for i in japan.index:
    print(i)
```
