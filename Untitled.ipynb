```python
import numpy as np
import pandas as pd 
import matplotlib.pyplot as plt 
%matplotlib inline
import seaborn as sns
import warnings
warnings.filterwarnings('ignore')
```

```python
# read csv
countries = pd.read_csv("World_Development_Indicators_2\world_bank_data.csv")

# data cleaning 
def get_country_data(country_name: str):
    result = countries[countries["Country Name"] == country_name].drop(columns = ["Country Name", "Country Code", "Series Code"])
    result = result.transpose()      # flip the table 
    result.columns = result.iloc[0]  # set columns to data variables 
    result = result[1:]
    result.index = pd.to_numeric(result.index.str[:4])  # alter the index (years) to yyyy format
    for col in result.columns:
        result[col] = pd.to_numeric(result[col], errors='coerce')   # convert everything from dtype 'object' to numeric 
    return result

# countries / aggregate data 
world = get_country_data("World")
japan = get_country_data("Japan")
united_states = get_country_data("United States")
china = get_country_data("China")
india = get_country_data("India")
low_income = get_country_data("Low income")
middle_income = get_country_data("Middle income")
high_income = get_country_data("High income")
```

```python
series = ["Population ages 0-14, total", "Population ages 15-64, total", "Population ages 65 and above, total"]

# create population demography plots
def plot_population_ages(data, title: str):
    data = data[series] / 1000000 # population in millions
    sns.lineplot(data)
    plt.ylim(0, data.max().max()+20)
    plt.title(title)
    plt.xlabel("Year") 
    plt.ylabel("Population (millions)") 
    plt.legend(title="Demographic Groups", loc="upper left", prop={'size': 8})

plt.figure(figsize=(10, 8))

# create graphs for different subplots 
plt.subplot(2, 2, 1)
plot_population_ages(united_states[:], "Age demography - United States")
plt.subplot(2, 2, 2)
plot_population_ages(japan[:], "Age demography - Japan")
plt.subplot(2, 2, 3)
plot_population_ages(china[:], "Age demography - China")
plt.subplot(2, 2, 4)
plot_population_ages(world[:], "Age demography - world")

plt.xticks(range(1960, 2030, 10))
plt.suptitle("Age Demography across different countries")

plt.tight_layout()  
plt.show()
```

```python
series = ["Population ages 0-14, total", "Population ages 15-64, total", "Population ages 65 and above, total", "Population, total"]

def get_dependency_ratio(data):
    data["Num dependents"] = data["Population ages 0-14, total"] + data["Population ages 65 and above, total"]
    # data["Num dependents"] = data["Population ages 0-14, total"] + data["Population ages 65 and above, total"] + (japan["Unemployment, total (% of total labor force) (national estimate)"] / 100) * data["Population, total"]
    data["Dependency ratio"] = data["Num dependents"] / data["Population ages 15-64, total"]
    return data 
```

```python
def test_regplot(data):
    data = get_dependency_ratio(data)
    data["Exports"] = data["GDP (current US$)"] * (data["Exports of goods and services (% of GDP)"] / 100)
    
    plt.figure(figsize=(5, 4))
    sns.regplot( x=data["Exports"], y=data["Dependency ratio"] )
    plt.show()

count = 0
for x in [world, japan, united_states, china, india, low_income, middle_income, high_income]:
    print(count)
    count += 1
    test_regplot(x[:])
```

```python
def test_regplot(data):
    data = get_dependency_ratio(data)
    data["Imports"] = data["GDP (current US$)"] * (data["Imports of goods and services (% of GDP)"] / 100)
    data["Exports"] = data["GDP (current US$)"] * (data["Exports of goods and services (% of GDP)"] / 100)
    data["Net exports"] = data["Exports"] - data["Imports"]
    
    plt.figure(figsize=(5, 4))
    sns.regplot( x=data["Net exports"], y=data["Dependency ratio"] )
    plt.show()

count = 0
for x in [world, japan, united_states, china, india, low_income, middle_income, high_income]:
    print(count)
    count += 1
    test_regplot(x[:])
```
